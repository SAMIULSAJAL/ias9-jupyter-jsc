```python
%matplotlib widget

import matplotlib.pyplot as plt
import numpy as np

# Use the image from our previous experiment
positive_points = []
negative_points = []

fig, ax = plt.subplots(figsize=(7, 7))
ax.imshow(image, cmap="gray")
ax.set_title("Left-click: Positive | Right-click: Negative")

def onclick(event):
    if event.inaxes != ax:
        return

    x, y = event.xdata, event.ydata

    if event.button == 1:
        positive_points.append([x, y])
        ax.plot(x, y, "go", markersize=8)
        print(f"Positive: ({x:.1f}, {y:.1f})")

    elif event.button == 3:
        negative_points.append([x, y])
        ax.plot(x, y, "ro", markersize=8)
        print(f"Negative: ({x:.1f}, {y:.1f})")

    fig.canvas.draw_idle()

cid = fig.canvas.mpl_connect("button_press_event", onclick)

plt.show()
```
